```python
import json
```

```python
# Carregar o JSON com os posts
with open('dataset_instagram-post-scraper_2025-12-01_17-32-33-885.json', 'r', encoding='utf-8') as f:
    posts = json.load(f)
```

```python

# Extrair apenas as URLs
urls = [post['url'] for post in posts]

# Criar JSON no formato do Apify
apify_input = {
    "directUrls": urls,
    "includeNestedComments": False,
    "isNewestComments": False,
    "resultsLimit": 15
}
```

```python
# Salvar
with open('apify_comments_input.json', 'w', encoding='utf-8') as f:
    json.dump(apify_input, f, indent=4, ensure_ascii=False)

print(f"✅ JSON criado com {len(urls)} URLs")
```
