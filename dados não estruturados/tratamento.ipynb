```python
import json
import pandas as pd
```

```python

with open('dados/row/dataset_instagram-post-scraper_2025-12-01_17-32-33-885.json', 'r', encoding='utf-8') as f:
    posts = json.load(f)

# Carregar comentários
with open('dados/row/dataset_instagram-comment-scraper_2025-12-03_02-58-19-419.json', 'r', encoding='utf-8') as f:
    comentarios = json.load(f)

print(f"📊 Posts carregados: {len(posts)}")
print(f"💬 Comentários carregados: {len(comentarios)}")
```

```python

df_posts = pd.DataFrame(posts)

# DataFrame de comentários
df_comentarios = pd.DataFrame(comentarios)


# Renomear colunas dos posts para evitar conflito
df_posts_renamed = df_posts.rename(columns={
    'url': 'post_url',
    'timestamp': 'post_timestamp'
})

# Fazer merge: comentários LEFT JOIN posts
# Cada comentário terá os dados do post correspondente
df_merged = df_comentarios.merge(
    df_posts_renamed,
    left_on='postUrl',
    right_on='post_url',
    how='left'
)
```

```python
colunas_finais = [
    # Dados do POST
    'post_url',
    'caption',
    'ownerFullName',
    'ownerUsername',
    'likesCount',
    'commentsCount',
    'post_timestamp',
    'firstComment',
    
    # Dados do COMENTÁRIO
    'id',
    'text',
    'timestamp',
    'ownerUsername_comment',  
    'ownerProfilePicUrl'
]

# Renomear coluna do username do comentário para não conflitar
df_merged = df_merged.rename(columns={
    'ownerUsername_y': 'ownerUsername_comment',  # username de quem comentou
    'ownerUsername_x': 'post_ownerUsername',     # username do dono do post
    'ownerUsername': 'post_ownerUsername'
})

# Selecionar e ordenar colunas
colunas_disponiveis = [
    'post_url',
    'caption',
    'ownerFullName',
    'post_ownerUsername',
    'likesCount',
    'commentsCount',
    'post_timestamp',
    'firstComment',
    'id',
    'text',
    'timestamp',
    'ownerUsername',
    'ownerProfilePicUrl'
]

# Usar apenas colunas que existem
colunas_existentes = [col for col in colunas_disponiveis if col in df_merged.columns]
df_final = df_merged[colunas_existentes]

# Renomear para ficar mais claro
df_final = df_final.rename(columns={
    'post_url': 'post_url',
    'caption': 'post_caption',
    'ownerFullName': 'post_owner_full_name',
    'post_ownerUsername': 'post_owner_username',
    'likesCount': 'post_likes_count',
    'commentsCount': 'post_comments_count',
    'post_timestamp': 'post_timestamp',
    'firstComment': 'post_first_comment',
    'id': 'comment_id',
    'text': 'comment_text',
    'timestamp': 'comment_timestamp',
    'ownerUsername': 'comment_owner_username',
    'ownerProfilePicUrl': 'comment_owner_profile_pic'
})

```

```python
# ============================================================
# ESTATÍSTICAS
# ============================================================

print("\n" + "="*60)
print("📊 ESTATÍSTICAS DO MERGE")
print("="*60)
print(f"✅ Total de linhas no CSV final: {len(df_final)}")
print(f"📝 Posts com comentários: {df_final['post_url'].nunique()}")
print(f"💬 Total de comentários: {len(df_final)}")

# Posts sem comentários
posts_com_comentarios = set(df_comentarios['postUrl'].unique())
posts_total = set(df_posts['url'].unique())
posts_sem_comentarios = posts_total - posts_com_comentarios
print(f"📭 Posts SEM comentários: {len(posts_sem_comentarios)}")

# ============================================================
# SALVAR CSV
# ============================================================

output_file = 'dados/trusted/dataset_comentarios_com_posts.csv'
df_final.to_csv(output_file, index=False, encoding='utf-8')

print(f"\n✅ CSV salvo em: {output_file}")
print(f"📋 Colunas: {list(df_final.columns)}")

# Mostrar preview
print("\n" + "="*60)
print("👀 PREVIEW DOS DADOS (primeiras 3 linhas)")
print("="*60)
print(df_final.head(3).to_string())
```
